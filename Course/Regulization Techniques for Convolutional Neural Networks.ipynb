A Residual Connection in a CNN is an Architectural Shortcut that bypasses One or More Layers by Adding the Layer's Input to its Output

```python
import torch
import torch.nn as nn

class SimpleResNet(nn.Module):
    def __init__(self, num_classes=10):
        super(SimpleResNet, self).__init__()
        
        self.initial_conv = nn.Conv2d(3, 64, kernel_size=3, padding=1)
        self.relu = nn.ReLU(inplace=True)
        
        self.b1_conv1 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=True)
        self.b1_conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=True)
        
        self.b2_conv1 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=True)
        self.b2_conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=True)
        
        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.fc = nn.Linear(64, num_classes)

    def forward(self, x):
        x = self.initial_conv(x)
        x = self.relu(x)
        
        identity1 = x 
        out1 = self.b1_conv1(x)
        out1 = self.relu(out1)
        out1 = self.b1_conv2(out1)
        x = self.relu(out1 + identity1)  
        
        identity2 = x  
        out2 = self.b2_conv1(x)
        out2 = self.relu(out2)
        out2 = self.b2_conv2(out2)
        x = self.relu(out2 + identity2)  
        
        x = self.pool(x)
        x = torch.flatten(x, 1)
        x = self.fc(x)
        return x
```

Batch Normalization ( Batch Norm ) and Dropout are two Essential Techniques Used to Improve the Training & Generalization of CNN's

```python
import torch
import torch.nn as nn

class SimpleBatchNormDropoutNet(nn.Module):
    def __init__(self, num_classes=10):
        super(SimpleBatchNormDropoutNet, self).__init__()
        
        self.initial_conv = nn.Conv2d(3, 64, kernel_size=3, padding=1, bias=False)
        self.initial_bn = nn.BatchNorm2d(64)
        self.relu = nn.ReLU(inplace=True)
        
        self.b1_conv1 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b1_bn1 = nn.BatchNorm2d(64)
        self.b1_conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b1_bn2 = nn.BatchNorm2d(64)
        
        self.b2_conv1 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b2_bn1 = nn.BatchNorm2d(64)
        self.b2_conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b2_bn2 = nn.BatchNorm2d(64)
        
        self.dropout2d = nn.Dropout2d(p=0.2)
        self.dropout = nn.Dropout(p=0.5)
        
        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.fc = nn.Linear(64, num_classes)

    def forward(self, x):
        x = self.initial_conv(x)
        x = self.initial_bn(x)
        x = self.relu(x)
        
        x = self.b1_conv1(x)
        x = self.b1_bn1(x)
        x = self.relu(x)
        x = self.b1_conv2(x)
        x = self.b1_bn2(x)
        x = self.relu(x)  
        x = self.dropout2d(x)  
        
        x = self.b2_conv1(x)
        x = self.b2_bn1(x)
        x = self.relu(x)
        x = self.b2_conv2(x)
        x = self.b2_bn2(x)
        x = self.relu(x)  
        x = self.dropout2d(x) 
        
        x = self.pool(x)
        x = torch.flatten(x, 1)
        x = self.dropout(x)  
        x = self.fc(x)
        return x
```

Learning Rate Scheduling in a Convolutional Neural Network ( CNN ) is the Process of Dynamically Changing the Learning Rate

```python
import torch
import torch.nn as nn
import torch.optim as optim

class FullResNet(nn.Module):
    def __init__(self, num_classes=10):
        super(FullResNet, self).__init__()
        
        self.initial_conv = nn.Conv2d(3, 64, kernel_size=3, padding=1, bias=False)
        self.initial_bn = nn.BatchNorm2d(64)
        self.relu = nn.ReLU(inplace=True)
        
        self.b1_conv1 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b1_bn1 = nn.BatchNorm2d(64)
        self.b1_conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b1_bn2 = nn.BatchNorm2d(64)
        
        self.b2_conv1 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b2_bn1 = nn.BatchNorm2d(64)
        self.b2_conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1, bias=False)
        self.b2_bn2 = nn.BatchNorm2d(64)
        
        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.fc = nn.Linear(64, num_classes)

    def forward(self, x):
        x = self.initial_conv(x)
        x = self.initial_bn(x)
        x = self.relu(x)
        
        identity1 = x  
        out1 = self.b1_conv1(x)
        out1 = self.b1_bn1(out1)
        out1 = self.relu(out1)
        out1 = self.b1_conv2(out1)
        out1 = self.b1_bn2(out1)
        x = self.relu(out1 + identity1)  
        
        identity2 = x  
        out2 = self.b2_conv1(x)
        out2 = self.b2_bn1(out2)
        out2 = self.relu(out2)
        out2 = self.b2_conv2(out2)
        out2 = self.b2_bn2(out2)
        x = self.relu(out2 + identity2)  
        
        x = self.pool(x)
        x = torch.flatten(x, 1)
        x = self.fc(x)
        return x

if __name__ == "__main__":
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    
    model = FullResNet(num_classes=10).to(device)
    
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.CrossEntropyLoss()
    
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.5)
    
    num_epochs = 10
    
    for epoch in range(num_epochs):
        model.train()
        
        inputs = torch.randn(4, 3, 224, 224).to(device)
        labels = torch.randint(0, 10, (4,)).to(device)
        
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        scheduler.step()
        
        current_lr = scheduler.get_last_lr()
        l = current_lr[0]
        print(f"Epoch {epoch+1:02d} | Loss : {loss.item():.4f} | Active LR : {l}")
```
